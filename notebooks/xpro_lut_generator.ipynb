# X-Pro LUT Generator

Bake a cross-processing look into a **3D LUT** you can use in the GLSL viewer, Photoshop, Lightroom, Resolve, or OBS.

Two paths:
1. **From a preset/profile** — bake the parametric model to a `.cube`.
2. **From a scan pair** — learn the mapping from a normal + cross-processed version of the same frame (see also `color_shift_analyzer.ipynb`).


```python
# Make the tools/ package importable from the notebooks/ folder.
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), "..", "tools")))
import numpy as np
import matplotlib.pyplot as plt
import xpro_core as xc
print("presets:", xc.list_presets())
```

## Bake a preset to a .cube LUT

```python
table = xc.profile_to_cube(xc.PRESETS["c41_in_e6"], size=33)
xc.write_cube("c41_in_e6.cube", table, title="c41_in_e6")
print("wrote c41_in_e6.cube", table.shape)
```

## Export a tiled PNG LUT for the WebGL viewer

`shaders/digital_xpro_lut.frag` reads a tiled (Hald) PNG because WebGL1 has no 3D textures. Size must be a perfect square (use 64).

```python
table64 = xc.profile_to_cube(xc.PRESETS["c41_in_e6"], size=64)
tiled = xc.table_to_tiled(table64)
xc.save_image("c41_in_e6_lut.png", tiled)
plt.figure(figsize=(4,4)); plt.imshow(tiled); plt.title("tiled 64-cube LUT"); plt.axis("off")
print("wrote c41_in_e6_lut.png", tiled.shape)
```
